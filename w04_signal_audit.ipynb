### 1. Distributions

```
import pandas as pd

df = pd.read_csv("flyrank-machine-learning/data/raw/content_refresh_anonymized.csv")
df["is_declining"] = (df["impressions_last_30d"] < 0.8 * df["impressions_prev_30d"]).astype(int)
df[["ctr", "avg_position", "engagement_rate", "scroll_rate"]].describe()
```

### 2. Signal test

```
df.groupby("is_declining")[["ctr", "avg_position", "engagement_rate", "scroll_rate"]].mean()
```

Verdict: CONFIRMED — CTR is clearly lower for declining pages (0.32 vs 0.73). avg_position and engagement_rate show weaker separation — MIXED.

### 3. The flag-linked test

```
df.groupby("is_declining")["trend_direction"].value_counts()
```

### 4. What this means in practice
CTR alone separates declining from stable pages more cleanly than position does. A content team could use CTR drop as a fast manual sanity check before trusting the model ranking.
